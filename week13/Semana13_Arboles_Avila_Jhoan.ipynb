## 2. Demostración guiada (resuelta)
Construimos un árbol de una universidad. Observe que cada nodo tiene una **etiqueta** y un **identificador único**.

```python
universidad = Tree()
universidad.create_node("Universidad", "u")
universidad.create_node("Facultad de Ingeniería", "fi", parent="u")
universidad.create_node("Facultad de Educación", "fe", parent="u")
universidad.create_node("Ingeniería de Sistemas", "sis", parent="fi")
universidad.create_node("Ingeniería Industrial", "ind", parent="fi")
universidad.show()
print("Padre de Sistemas:", universidad.parent("sis").tag)
print("Hijos de Ingeniería:", [n.tag for n in universidad.children("fi")])
```

## Ejercicio 5. Caso aplicado: servicios universitarios (20 puntos)
Diseñe un árbol de navegación para un portal universitario con raíz **Portal**, al menos cuatro servicios principales y dos trámites por servicio.

Implemente una función `ruta_hasta_raiz(arbol, identificador)` que devuelva la ruta desde un trámite hasta la raíz (en orden raíz → trámite). Pruebe la función con dos trámites diferentes.

**Evidencia:** árbol visual y dos rutas completas.

```python
# TODO: implementar el ejercicio 5
portal = Tree()
portal.create_node("Portal", "p")

# Crear los cuatro servicios principales
portal.create_node("Académico", "aca", parent="p")
portal.create_node("Financiero", "fin", parent="p")
portal.create_node("Bienestar universitario", "bie", parent="p")
portal.create_node("Biblioteca", "bib", parent="p")

# Trámites del servicio Académico
portal.create_node("Inscripción de materias", "ins", parent="aca")
portal.create_node("Solicitud de certificados", "cer", parent="aca")

# Trámites del servicio Financiero
portal.create_node("Pago de matrícula", "pag", parent="fin")
portal.create_node("Solicitud de descuentos", "des", parent="fin")

# Trámites del servicio Bienestar Universitario
portal.create_node("Inscripción a deportes", "dep", parent="bie")
portal.create_node("Solicitud de apoyo psicológico", "psi", parent="bie")

# Trámites del servicio Biblioteca
portal.create_node("Préstamo de libros", "pre", parent="bib")
portal.create_node("Prestamo de aulas", "ren", parent="bib")

print("========== Arbol Portal ==========")
portal.show()


# Busqueda de ruta
def ruta_hasta_raiz(arbol, identificador):
    ruta = []
    nodo = arbol.get_node(identificador)

    if nodo is None:
        return []

    # Recorrer desde el nodo hasta la raíz
    while nodo is not None:
        ruta.append(nodo.tag)
        padre = arbol.parent(nodo.identifier)
        if padre is None:
            break
        nodo = padre
    ruta.reverse()

    return ruta


# Ejecutar funcion
print("\n========== Rutas de navegacion ==========")
ruta1 = ruta_hasta_raiz(portal, "cer")
print("Ruta 1 - Solicitud de certificados:")
print(" -> ".join(ruta1))

ruta2 = ruta_hasta_raiz(portal, "pag")
print("\nRuta 2 - Pago de matrícula:")
print(" -> ".join(ruta2))
```
